### 1.1: Load and Review Your Data


````python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

# Edit for dataset options 2 or 3

# Option 1: Metallic glass formation database (you choose option 2 or 3 below)
with open('../data_files/database/glass.json', 'r') as f:
    glass_data = json.load(f)

# Option 2: Steel yield strength data
# with open('../data_files/database/steels_yield.json', 'r') as f:
#    steel_data = json.load(f)

# Option 3: Metal classification data
# with open('../data_files/database/matbench_expt_is_metal.json', 'r') as f:
#    metal_data = json.load(f)

# Process your data (it has a specific structure)
variable_1 = []
variable_2 = []
for item in glass_data['data']: # change glass_data to your data name
    variable_1.append(item[0])
    variable_2.append(item[1])

# Create DataFrame
data = pd.DataFrame({
    'Composition': variable_1, # change "Composition" to your data name
    'Glass_Forming': variable_2 # change "Glass_Forming" to your data name
})

print(f"Dataset loaded: {data.shape}")
print(f"Columns: {list(data.columns)}")
print("\nFirst few rows:")
data.head()
````

### 1.2: Quick Data Analysis

**Task**: Perform a quick analysis to identify key insights for your dashboard


````python
# === Quick Data Analysis ===

# Basic statistics
print("=== DATASET OVERVIEW ===")
print(f"Shape: {data.shape}")
print(f"Columns: {list(data.columns)}")
print(f"Data types:\n{data.dtypes}")

# Check for missing values
print("\n=== MISSING VALUES ===")
print(data.isnull().sum())

# Basic descriptive statistics
print("\n=== DESCRIPTIVE STATISTICS ===")
print(data.describe())

# Categorical variable analysis
categorical_cols = data.select_dtypes(include=['object', 'bool']).columns
if len(categorical_cols) > 0:
    print("\n=== CATEGORICAL VARIABLES ===")
    for col in categorical_cols:
        print(f"\n{col}:")
        print(data[col].value_counts().head())

````

### 1.3: Feature Engineering for Dashboard

**Task**: Based on the available data, you will usually create additional features (data columns) that will be useful for visualization. Here are some examples for glass_data:

#### 1.3.1 📏 Composition Length
- **Why**: Composition strings have different lengths proportional to the number of elements. Using the string length is simple and doesn't require code to parse the string into the different elements.
- **Con**: It is more accurate to count the number of elements. 
- **Use**: Show distribution of composition complexity  
- **Dashboard value**: Histogram of composition lengths

#### 1.3.2. 🔢 Number of Elements
- **Why**: Different compositions have different numbers of elements
- **Con**: Will need to break up (parse) the composition string into the different elements.
- **Use**: Analyze relationship between complexity and glass formation
- **Dashboard value**: Scatter plot or grouped analysis

#### 1.3.3. 💡 Other Possible Features
- Element classification analysis - Compare the ratio of metallic vs non-metallic elements in each composition
- Element size diversity - Analyze the distribution of atomic radii or element sizes within compositions
- Composition balance analysis - Examine the relative proportions of different elements in each alloy
- Representing data on Ternary diagram for 3 component systems?
- Anything else?

#### 1.3.4. 🎯 Decision Time
Based on our data overview above, what features do we want to include on our dashboard? 

**Keeping things simple**: Let's add composition length and element count as these are the easiest to calculate for this simple demo. 


````python
# Add composition length column (useful for complexity analysis)
data['Composition_Length'] = data['Composition'].str.len()

# Add number of elements column (useful for multi-element analysis)
import re # regular expressions package - used to analyze text strings
data['num_elements'] = data['Composition'].apply(lambda x: len(re.findall(r'[A-Z][a-z]?', x)))
# the pattern r'[A-Z][a-z]?' in the line above matches an upper case letter 
# followed by an optional lower case letter such as Cu, Al, Fe

# Check the new features
print("✅ Added new features:")
print(f"   - Composition_Length: range {data['Composition_Length'].min()} to {data['Composition_Length'].max()}")
print(f"   - num_elements: range {data['num_elements'].min()} to {data['num_elements'].max()}")

print(f"\nUpdated dataset shape: {data.shape}")
print(f"Updated columns: {list(data.columns)}")

print("\nFirst few rows:")
data.head()

# Export the updated data to a CSV file for dashboard use
data.to_csv('dashboard_data.csv', index=False)
print("✅ Data exported to dashboard_data.csv")
````

### Common Issues & Solutions

**❌ "streamlit is not recognized as a command" or "ModuleNotFoundError"**
- **Solution**: Install Streamlit first: `pip install streamlit`
---
**❌ "No such file or directory"**
- **Solution**: Make sure you're in the correct directory where your `.py` file is located
---
**❌ "dashboard visuals not displaying in grid or not where you want them"**
- **Problem**: When you paste AI-generated code into the `with col1:` sections, it needs proper indentation
- **Solution**: Make sure your code is indented relative to the `with col1:` statement:

  ```python
  with col1:
      st.subheader("📊 Visualization 1")
      # ✅ CORRECT - code is indented under the 'with col1:' statement
      fig = px.histogram(data, x='Composition_Length', title='Distribution')
      st.plotly_chart(fig, use_container_width=True)
  ```

  ```python
  with col1:
      st.subheader("📊 Visualization 1")
  fig = px.histogram(data, x='Composition_Length', title='Distribution')  # ❌ WRONG - not indented
  st.plotly_chart(fig, use_container_width=True)  # ❌ WRONG - not indented
  ```
---
**❌ Charts are full width and don't fit in columns**
- **Solution**: Add `use_container_width=True` to your `st.plotly_chart()` calls:
  ```python
  st.plotly_chart(fig, use_container_width=True)
  ```
**Also make sure you are using proper indentations**
