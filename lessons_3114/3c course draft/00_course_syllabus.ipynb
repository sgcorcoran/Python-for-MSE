## Course Description

MSE 3114 is an advanced, project-based course that introduces materials science students to the transformative power of artificial intelligence in research and analysis. Building upon the Python programming foundation from MSE 2114, this course explores how AI tools can enhance every aspect of materials science work—from experimental design and data analysis to literature review and collaboration.

Students will learn to integrate AI assistants (ChatGPT, Claude, local LLMs), automated coding tools (GitHub Copilot), and modern data science techniques to solve real materials science problems. The course emphasizes practical application, with hands-on projects that demonstrate how AI can accelerate research, improve accuracy, and enable new types of analysis that would be impractical or impossible with traditional methods alone.

---

## Learning Objectives

By the end of this course, students will be able to:

### Technical Skills
- **AI Integration**: Effectively use AI tools for materials science research, analysis, and problem-solving
- **Modern Data Stack**: Implement Polars, DuckDB, and Streamlit for efficient data processing and visualization
- **Advanced Analysis**: Apply machine learning, statistical analysis, and optimization techniques to materials data
- **Image Processing**: Perform automated microstructural analysis using computer vision and AI techniques
- **Experimental Design**: Use AI-enhanced design of experiments for parameter optimization

### Research Competencies
- **Literature Review**: Conduct AI-assisted literature reviews and research synthesis
- **Data Management**: Design robust data pipelines with validation and version control
- **Model Development**: Create, validate, and optimize predictive models for materials properties
- **Quality Control**: Implement AI-enhanced quality assurance and defect detection systems
- **Collaboration**: Use AI tools to enhance research collaboration and knowledge sharing

### Professional Development
- **Critical Thinking**: Evaluate AI-generated results and identify potential biases or errors
- **Ethical Considerations**: Understand data privacy, AI limitations, and responsible AI use in research
- **Communication**: Present technical results using interactive dashboards and AI-enhanced visualizations
- **Innovation**: Develop creative approaches to materials science problems using AI augmentation
