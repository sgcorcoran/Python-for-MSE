### I. Anaconda Python Environment Setup

#### Anaconda Update (if you already have Anaconda installed)

**Step 1: Open Anaconda Prompt**
- Click the **Start** button (Windows icon)
- Type "Anaconda Prompt" 
- Click on **"Anaconda Prompt (anaconda3)"** or **"Anaconda PowerShell"**

**Step 2: Update Conda (the package manager)**
In the Anaconda Prompt window, type this command and press **Enter**:
```bash
conda update conda
```
- When prompted, type **y** and press **Enter** to confirm
- Wait for the update to complete

**Step 3: Update All Packages**
After Step 2 completes, type this command and press **Enter**:
```bash
conda update --all
```
- When prompted, type **y** and press **Enter** to confirm
- This may take several minutes - be patient!

**Step 4: Clean Up (Optional but Recommended)**
After all updates complete, type this command and press **Enter**:
```bash
conda clean --all
```
- When prompted, type **y** and press **Enter** to confirm

**Step 5: Verify the Update**
Type this command to check your conda version:
```bash
conda --version
```

**Step 6: Close the Window**
- Type **exit** and press **Enter**
- Or simply close the Anaconda Prompt window



##### What This Does
- **Step 2**: Updates the conda package manager itself
- **Step 3**: Updates all your installed packages to their latest compatible versions
- **Step 4**: Removes old package files to free up disk space

##### If Something Goes Wrong
- Don't panic - your old packages are still there
- Close the Anaconda Prompt and try again
- If problems persist, ask your instructor for help
