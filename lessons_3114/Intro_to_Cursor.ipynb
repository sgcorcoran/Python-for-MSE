# Introduction to Cursor: AI-Powered Coding Assistant

## What is Cursor?
Cursor is a modern code editor that integrates AI assistance to help you write, debug, and understand code more effectively. Think of it as having a coding partner who can help you solve problems and learn new concepts.

## Why We're Using Both Tools

### Jupyter Notebooks (Familiar)
- **Interactive coding** with cells that run independently
- **Data visualization** and analysis workflows
- **Documentation** alongside code
- **Export options** for reports and presentations

### Cursor (New Addition)
- **AI code assistance** - get help when you're stuck
- **Better code editing** experience
- **Integrated terminal** and file management
- **Git integration** for version control
- **Multiple file types** support (Python, Markdown, etc.)

## How This Helps You

1. **Faster Problem Solving** - Ask AI for help with errors or concepts
2. **Better Code Quality** - AI can suggest improvements and best practices
3. **Learning Support** - Get explanations when you don't understand something
4. **Efficiency** - Spend less time debugging and more time learning

## What to Expect
- You'll still use Jupyter notebooks for data analysis and visualization
- Cursor will be your primary code editor for writing Python scripts and functions
- AI assistance is available whenever you need help
- Both tools work together seamlessly

## Installing Cursor: Step-by-Step

### Step 1: Download
1. **Go to** [cursor.sh](https://cursor.sh)
2. **Click "Download"** button
3. **Choose your operating system** (Windows, Mac, or Linux)
4. **Wait for download** to complete

### Step 2: Install
1. **Run the installer** (double-click the downloaded file)
2. **Follow installation prompts** (accept defaults for now)
3. **Wait for installation** to complete
4. **Click "Finish"** when done

### Step 3: First Launch
1. **Open Cursor** from your Start Menu/Applications
2. **Sign in** with your GitHub account (recommended) or skip for now
3. **Choose your theme** (light or dark - you can change later)
4. **Complete setup** and you'll see the welcome screen

### Step 4: Install Jupyter Extension
1. **Press Ctrl + Shift + X** (Windows/Linux) or **Cmd + Shift + X** (Mac)
2. **Search for "Jupyter"**
3. **Install "Jupyter"** extension by Microsoft
4. **Restart Cursor** when prompted

### Step 5: Test Your Setup
1. **Create a new file** with `.ipynb` extension
2. **Verify Jupyter interface** appears
3. **Test a simple code cell**:
   ```python
   print("Hello, Cursor!")
   ```

## What You Should See
- **Dark/light theme** based on your choice
- **File explorer** on the left side
- **Editor area** in the center
- **Terminal** at the bottom (can be toggled)
- **Jupyter notebook interface** when working with .ipynb files

### Troubleshooting
- **If Jupyter doesn't work**: Make sure you have Python and Jupyter installed via Anaconda
- **If themes look wrong**: You can change them in Settings (Ctrl/Cmd + ,)
- **If extensions don't load**: Try restarting Cursor

## Creating Your First Project in Cursor

### Step 1: Open a Project Folder
1. **Click "Open Folder"** from the welcome screen, or
2. **Press Ctrl + K, Ctrl + O** (Windows/Linux) or **Cmd + K, Cmd + O** (Mac)
3. **Navigate to** your course folder (e.g., `Python-for-MSE`)
4. **Click "Select Folder"**

### Step 2: Create a New Jupyter Notebook
1. **Right-click** in the file explorer (left panel)
2. **Select "New File"**
3. **Name it** `test_cursor.ipynb` (the .ipynb extension is important!)
4. **Press Enter**

### Step 3: Test Basic Functionality
1. **Click in the first cell** (it should be a code cell by default)
2. **Type this simple code**:
   ```python
   print("Hello from Cursor!")
   ```
3. **Press Shift + Enter** to run the cell
4. **You should see** the output below the cell

### Step 4: Test Markdown Cells
1. **Click on a new cell** (or press **B** to add a cell below)
2. **Press M** to convert it to a markdown cell
3. **Type some markdown**:
   ```markdown
   # My First Notebook in Cursor
   This is a **markdown cell** with *formatting*.
   ```
4. **Press Shift + Enter** to render it

## Exploring AI Assistance

### Your First AI Interaction
1. **Press Ctrl + K** (Windows/Linux) or **Cmd + K** (Mac)
2. **Type a question** like: "How do I create a list in Python?"
3. **Press Enter** and see the AI response
4. **Try asking**: "Explain what this code does" while selecting some code

### AI Features to Try
- **Code explanation**: Select code and ask "What does this do?"
- **Error help**: Paste an error message and ask "How do I fix this?"
- **Concept questions**: Ask "What is a DataFrame in pandas?"
- **Code improvement**: Ask "How can I make this code more efficient?"
