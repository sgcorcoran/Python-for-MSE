# MSE 3114 Mid-Course Evaluation Survey Analysis

This notebook analyzes the survey results according to the SURVEY_ANALYSIS_GUIDE.md framework.


## Load the Survey Data


```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load the survey data
survey_file = Path("MSE 3114_ Mid-Course Evaluation Survey (Anonymous)(1-22).xlsx")
df = pd.read_excel(survey_file)

# Display basic info about the data
print(f"Total responses: {len(df)}")
print(f"\nColumn names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col[:80]}...")
    
print(f"\nFirst few rows:")
df.head()
```

## Key Questions Analysis

### Question 1: Python Comfort Level (1-5 scale)


```python
# Analyze Q1: Python comfort level
q1_col = 'Technical Comfort Level (Multiple Choice - Single Answer)\nOn a scale of 1-5, how comfortable are you with Python programming now?'

q1_counts = df[q1_col].value_counts().sort_index()
q1_pct = (q1_counts / len(df) * 100).round(1)

print("Question 1: Python Comfort Level")
print("=" * 50)
for val, count in q1_counts.items():
    pct = q1_pct[val]
    print(f"{val}: {count} students ({pct}%)")

# Identify struggling vs confident students
struggling = df[df[q1_col].isin([1, 2])].shape[0]
confident = df[df[q1_col].isin([4, 5])].shape[0]
print(f"\nStruggling (1-2): {struggling} ({struggling/len(df)*100:.1f}%)")
print(f"Confident (4-5): {confident} ({confident/len(df)*100:.1f}%)")
```

### Question 2: Course Difficulty


```python
# Analyze Q2: Course difficulty
q2_col = 'Course Difficulty (Multiple Choice - Single Answer)\nHow would you rate the overall difficulty of lessons 01-04?'

q2_counts = df[q2_col].value_counts()
q2_pct = (q2_counts / len(df) * 100).round(1)

print("Question 2: Course Difficulty")
print("=" * 50)
for val, count in q2_counts.items():
    pct = q2_pct[val]
    print(f"{val}: {count} students ({pct}%)")
```

### Question 3: Pacing (Critical for Content Adjustment)


```python
# Analyze Q3: Pacing
q3_col = 'Pacing (Multiple Choice - Single Answer)\nHow would you rate the pace of lessons 01-04?'

q3_counts = df[q3_col].value_counts()
q3_pct = (q3_counts / len(df) * 100).round(1)

print("Question 3: Course Pacing")
print("=" * 50)
for val, count in q3_counts.items():
    pct = q3_pct[val]
    print(f"{val}: {count} students ({pct}%)")

# Check if too fast
too_fast_options = ['too fast', 'way too fast', 'fast']
too_fast_count = df[q3_col].str.lower().str.contains('too fast|way too fast', case=False, na=False).sum()
print(f"\nToo Fast: {too_fast_count} ({too_fast_count/len(df)*100:.1f}%)")
if too_fast_count / len(df) > 0.4:
    print("⚠️  WARNING: >40% find pacing too fast - consider reducing content")
```

### Question 4: Time Investment


```python
# Analyze Q4: Time spent per week
q4_col = 'Time Investment (Multiple Choice - Single Answer)\nOn average, how many hours per week did you spend on this course?'

q4_counts = df[q4_col].value_counts()
q4_pct = (q4_counts / len(df) * 100).round(1)

print("Question 4: Time Spent Per Week")
print("=" * 50)
for val, count in q4_counts.items():
    pct = q4_pct[val]
    print(f"{val}: {count} students ({pct}%)")

# Check high time (>6 hours)
high_time_count = df[q4_col].astype(str).str.lower().str.contains('6|7|more', case=False, na=False).sum()
print(f"\n6+ hours: {high_time_count} ({high_time_count/len(df)*100:.1f}%)")
if high_time_count / len(df) > 0.5:
    print("⚠️  WARNING: >50% spending 6+ hours - consider streamlining")
```

### Question 5: Most Helpful Aspects


```python
# Analyze Q5: Most helpful aspects - multi-select question
# Microsoft Forms exports as a single column with semicolon-separated values
q5_col = 'Most Helpful Aspects (Multiple Choice - Multiple Answers, up to 3)\nSelect up to 3 things that helped you learn most:'

print("Question 5: Most Helpful Aspects")
print("=" * 50)

# Split by semicolon (Microsoft Forms uses semicolons for multi-select)
all_helpful = []
for val in df[q5_col].dropna():
    if isinstance(val, str):
        # Split by semicolon and strip whitespace, filter out empty strings
        helpful = [c.strip() for c in val.split(';') if c.strip()]
        all_helpful.extend(helpful)

helpful_counts = pd.Series(all_helpful).value_counts()
helpful_pct = (helpful_counts / len(df) * 100).round(1)

print("\nMost Helpful Aspects (sorted by frequency):")
for aspect, count in helpful_counts.items():
    pct = helpful_pct[aspect]
    print(f"{aspect}: {count} ({pct}%)")
```

### Question 6: Challenges (Critical - Identifies Gaps)


```python
# Analyze Q6: Challenges - this is a multi-select question
# Microsoft Forms exports as a single column with semicolon-separated values
q6_col = 'Most Challenging Aspects (Multiple Choice - Multiple Answers, up to 3)\nSelect up to 3 things that were most challenging:'

print("Question 6: Challenges")
print("=" * 50)

# Split by semicolon (Microsoft Forms uses semicolons for multi-select)
all_challenges = []
for val in df[q6_col].dropna():
    if isinstance(val, str):
        # Split by semicolon and strip whitespace, filter out empty strings
        challenges = [c.strip() for c in val.split(';') if c.strip()]
        all_challenges.extend(challenges)

challenge_counts = pd.Series(all_challenges).value_counts()
challenge_pct = (challenge_counts / len(df) * 100).round(1)

print("\nChallenge Summary (sorted by frequency):")
for challenge, count in challenge_counts.items():
    pct = challenge_pct[challenge]
    print(f"{challenge}: {count} ({pct}%)")
    if pct > 40:
        print(f"  ⚠️  HIGH (>40%)")
    elif pct > 30:
        print(f"  ⚠️  MEDIUM (>30%)")
```

### Question 8: AI Tool Effectiveness


```python
# Analyze Q8: AI tool helpfulness
q8_col = 'AI Tool Effectiveness (Multiple Choice - Single Answer)\nHow helpful have AI tools been for your learning?'

q8_counts = df[q8_col].value_counts()
q8_pct = (q8_counts / len(df) * 100).round(1)

print("Question 8: AI Tool Helpfulness")
print("=" * 50)
for val, count in q8_counts.items():
    pct = q8_pct[val]
    print(f"{val}: {count} students ({pct}%)")

# Check if not helpful
not_helpful = df[q8_col].str.lower().str.contains('not helpful|somewhat', case=False, na=False).sum()
print(f"\nNot/Somewhat Helpful: {not_helpful} ({not_helpful/len(df)*100:.1f}%)")
if not_helpful / len(df) > 0.3:
    print("⚠️  WARNING: >30% find AI tools not helpful - add AI usage examples")
```

### Question 9: AI Challenges (Important for Lesson 05)


```python
# Analyze Q9: AI challenges - multi-select
# Microsoft Forms exports as a single column with semicolon-separated values
q9_col = 'AI Tool Challenges (Multiple Choice - Multiple Answers)\nWhat challenges have you faced using AI tools? (Select all that apply)'

print("Question 9: AI Challenges")
print("=" * 50)

# Split by semicolon (Microsoft Forms uses semicolons for multi-select)
all_challenges = []
for val in df[q9_col].dropna():
    if isinstance(val, str):
        # Split by semicolon and strip whitespace, filter out empty strings
        challenges = [c.strip() for c in val.split(';') if c.strip()]
        all_challenges.extend(challenges)

challenge_counts = pd.Series(all_challenges).value_counts()
challenge_pct = (challenge_counts / len(df) * 100).round(1)

print("\nAI Challenge Summary (sorted by frequency):")
for challenge, count in challenge_counts.items():
    pct = challenge_pct[challenge]
    print(f"{challenge}: {count} ({pct}%)")
    if pct > 40:
        print(f"  ⚠️  HIGH - Address in Lesson 05")
    elif pct > 30:
        print(f"  ⚠️  MEDIUM - Consider for Lesson 05")
```

### Question 10: Concerns About Upcoming Lessons (CRITICAL for Lesson Tweaks)


```python
# Analyze Q10: Concerns - CRITICAL for lesson adjustments
# Microsoft Forms exports as a single column with semicolon-separated values
q10_col = 'Concerns About Lessons 05-07 (Multiple Choice - Multiple Answers)\nWhat are you most concerned about for the remaining lessons? (Select all that apply)'

print("Question 10: Concerns About Upcoming Lessons")
print("=" * 50)

# Split by semicolon (Microsoft Forms uses semicolons for multi-select)
all_concerns = []
for val in df[q10_col].dropna():
    if isinstance(val, str):
        # Split by semicolon and strip whitespace, filter out empty strings
        concerns = [c.strip() for c in val.split(';') if c.strip()]
        all_concerns.extend(concerns)

concern_counts = pd.Series(all_concerns).value_counts()
concern_pct = (concern_counts / len(df) * 100).round(1)

print("\nTop Concerns (sorted by frequency):")
for concern, count in concern_counts.items():
    pct = concern_pct[concern]
    print(f"\n{concern}: {count} ({pct}%)")
    
    # Actionable recommendations based on guide
    concern_lower = concern.lower()
    if 'validation' in concern_lower or 'ai validation' in concern_lower:
        if pct > 40:
            print(f"  → HIGH PRIORITY: Add more scaffolding to Lesson 05")
        elif pct > 30:
            print(f"  → MEDIUM: Consider more scaffolding for Lesson 05")
    elif 'workflow' in concern_lower or 'integrated' in concern_lower:
        if pct > 40:
            print(f"  → HIGH PRIORITY: Simplify Lesson 06")
        elif pct > 30:
            print(f"  → MEDIUM: Consider simplifying Lesson 06")
    elif 'capstone' in concern_lower:
        if pct > 40:
            print(f"  → HIGH PRIORITY: Provide more structure in Lesson 07")
        elif pct > 30:
            print(f"  → MEDIUM: Consider adding structure to Lesson 07")
    elif 'time' in concern_lower or 'pacing' in concern_lower:
        if pct > 40:
            print(f"  → HIGH PRIORITY: Reduce content or extend deadlines")
    elif 'python' in concern_lower or 'coding' in concern_lower:
        if pct > 30:
            print(f"  → Consider Python review resources")
```

```python
# Analyze Q11: Support needs - DIRECTLY tells you what to add
# Microsoft Forms exports as a single column with semicolon-separated values
q11_col = 'Support Needs (Multiple Choice - Multiple Answers, up to 2)\nWhat additional support would be most helpful for lessons 05-07? (Select up to 2)'

print("Question 11: Support Needs")
print("=" * 50)

# Split by semicolon (Microsoft Forms uses semicolons for multi-select)
all_needs = []
for val in df[q11_col].dropna():
    if isinstance(val, str):
        # Split by semicolon and strip whitespace, filter out empty strings
        needs = [c.strip() for c in val.split(';') if c.strip()]
        all_needs.extend(needs)

need_counts = pd.Series(all_needs).value_counts()
need_pct = (need_counts / len(df) * 100).round(1)

print("\nTop Support Needs (sorted by frequency):")
for need, count in need_counts.items():
    pct = need_pct[need]
    print(f"\n{need}: {count} ({pct}%)")
    
    # Actionable recommendations based on guide
    need_lower = need.lower()
    if 'python' in need_lower and ('basic' in need_lower or 'review' in need_lower):
        if pct > 50:
            print(f"  → HIGH PRIORITY: Add Python review sections to upcoming lessons")
        elif pct > 30:
            print(f"  → MEDIUM: Consider Python review resources")
    elif 'validation' in need_lower and ('step' in need_lower or 'example' in need_lower):
        if pct > 50:
            print(f"  → HIGH PRIORITY: Enhance Lesson 05 examples with step-by-step validation")
        elif pct > 30:
            print(f"  → MEDIUM: Add more AI validation examples")
    elif 'example' in need_lower or 'worked' in need_lower:
        if pct > 50:
            print(f"  → HIGH PRIORITY: Add complete workflow walkthroughs")
        elif pct > 30:
            print(f"  → MEDIUM: Add more worked examples")
    elif 'prompt' in need_lower:
        if pct > 30:
            print(f"  → Add AI prompt templates and examples")
    elif 'video' in need_lower:
        if pct > 30:
            print(f"  → Consider adding video tutorials")
    elif 'statistical' in need_lower or 'concept' in need_lower:
        if pct > 30:
            print(f"  → Consider clearer explanations of statistical concepts")
```
