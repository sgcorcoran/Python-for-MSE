## Mini-Lecture: AI-Assisted Data Processing Strategy (~20-25 min)

### The AI-Augmented Data Processing Workflow

Traditional data processing follows a linear path: **Load → Clean → Analyze → Visualize**. AI augmentation transforms this into an **iterative, consultative process** where AI helps with decision-making at each step.

**Core Principles for AI-Assisted Data Processing:**

1. **Problem-First Approach** — Start with the materials science question, not the tool
2. **AI Consultation** — Use AI to explore options and trade-offs before committing
3. **Iterative Refinement** — Test AI suggestions, gather feedback, improve approach
4. **Validation-Driven** — Always verify AI recommendations with data and domain knowledge

### AI Prompt Patterns for Data Processing

### Materials Science Data Processing Challenges

**Common Issues in Materials Datasets:**
- **Mixed data types**: Composition percentages, temperature ranges, property measurements
- **Experimental uncertainty**: Measurement errors, instrument precision, sample variability
- **Unit conversions**: SI vs. imperial, different temperature scales, pressure units
- **Missing data**: Failed experiments, incomplete measurements, equipment issues
- **Large file sizes**: High-resolution microscopy, extensive property databases
- **Complex relationships**: Non-linear property correlations, phase transformations

**AI Can Help With:**
- **Tool selection** based on data characteristics and analysis goals
- **Workflow optimization** for specific materials science tasks
- **Error detection** and data validation strategies
- **Performance tuning** for large materials datasets
- **Visualization design** for materials property relationships

**AI Limitations to Watch For:**
- **Over-optimization**: May suggest complex solutions for simple problems
- **Tool bias**: May favor certain libraries without considering alternatives
- **Performance estimates**: May not account for your specific hardware/constraints
- **Domain knowledge gaps**: May miss materials-specific considerations

**Polars** — Fast, memory-efficient alternative to pandas

- **Materials applications**: Property databases, composition analysis, statistical processing
- **AI can help with**: Syntax conversion, optimization strategies, performance tuning

**Dask** — Parallel computing for larger-than-memory datasets
- **Best for**: Very large datasets (>1GB), distributed computing, complex workflows
- **Materials applications**: High-resolution microscopy, extensive property databases
- **AI can help with**: Workflow design, resource allocation, optimization strategies

**Streamlit** — Interactive dashboard creation
- **Best for**: Research communication, interactive data exploration, stakeholder presentations
- **Materials applications**: Property visualization, composition optimization, results sharing
- **AI can help with**: UI design, functionality planning, deployment strategies

````python
# === Basic Streamlit Dashboard Template ===
# Use AI to help you customize this template for your specific dataset

import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go

# Dashboard title and description
st.title("Materials Science Data Dashboard")
st.write("Interactive exploration of alloy composition-property relationships")

# Sidebar controls
st.sidebar.header("Filter Controls")

# Example filter controls (customize based on your dataset)
composition_range = st.sidebar.slider(
    "Aluminum Content Range (%)",
    min_value=float(data_large['Al_content'].min()),
    max_value=float(data_large['Al_content'].max()),
    value=(float(data_large['Al_content'].min()), float(data_large['Al_content'].max()))
)

strength_range = st.sidebar.slider(
    "Tensile Strength Range (MPa)",
    min_value=float(data_large['Tensile_Strength_MPa'].min()),
    max_value=float(data_large['Tensile_Strength_MPa'].max()),
    value=(float(data_large['Tensile_Strength_MPa'].min()), float(data_large['Tensile_Strength_MPa'].max()))
)

# Filter data based on controls
filtered_data = data_large[
    (data_large['Al_content'] >= composition_range[0]) &
    (data_large['Al_content'] <= composition_range[1]) &
    (data_large['Tensile_Strength_MPa'] >= strength_range[0]) &
    (data_large['Tensile_Strength_MPa'] <= strength_range[1])
]

# Display filtered data info
st.write(f"Showing {len(filtered_data)} alloys out of {len(data_large)} total")

# Create visualizations
col1, col2 = st.columns(2)

with col1:
    # Scatter plot: Composition vs. Properties
    fig1 = px.scatter(
        filtered_data,
        x='Al_content',
        y='Tensile_Strength_MPa',
        color='Yield_Strength_MPa',
        size='Elongation_Percent',
        hover_data=['Cu_content', 'Mg_content'],
        title="Aluminum Content vs. Tensile Strength",
        labels={'Al_content': 'Aluminum Content (%)', 'Tensile_Strength_MPa': 'Tensile Strength (MPa)'}
    )
    st.plotly_chart(fig1, use_container_width=True)

with col2:
    # Histogram: Property distribution
    fig2 = px.histogram(
        filtered_data,
        x='Tensile_Strength_MPa',
        nbins=20,
        title="Tensile Strength Distribution",
        labels={'Tensile_Strength_MPa': 'Tensile Strength (MPa)', 'count': 'Number of Alloys'}
    )
    st.plotly_chart(fig2, use_container_width=True)

# Summary statistics
st.subheader("Summary Statistics")
summary_stats = filtered_data[['Al_content', 'Cu_content', 'Mg_content', 
                              'Tensile_Strength_MPa', 'Yield_Strength_MPa', 'Elongation_Percent']].describe()
st.dataframe(summary_stats)

# Instructions for running
st.info("""
**To run this dashboard:**
1. Save this code as `dashboard.py`
2. Install Streamlit: `pip install streamlit`
3. Run: `streamlit run dashboard.py`
4. Open the URL shown in your terminal
""")
````
