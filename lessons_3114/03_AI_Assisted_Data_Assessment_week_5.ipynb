## Week 5: AI-Assisted Data Assessment

### Mini-Lecture: AI Prompt Patterns for Data Processing (15 min)

**Learning Objectives**:
- Understand the structure and components of effective AI prompts
- Master the Data Assessment prompt pattern for materials science datasets
- Learn how to iterate on AI assessments for deeper insights

**Key Concepts**:

#### 1. The RAG-IT Framework for AI Prompts

Effective AI prompts follow the **RAG-IT** framework:

- **R**ole: Define the AI's expertise and perspective
- **A**udience: Specify who will use the AI's output
- **G**oal: State the primary objective clearly
- **I**nputs: Provide specific data and context
- **T**ask: Describe the exact work to be done

**Additional Components** (often included for better results):
- **CONSTRAINTS**: Specify word limits, time estimates, tool options, or other limitations
- **EVAL CRITERIA**: Define what makes a successful response (specific, actionable, etc.)

#### 2. Data Assessment Prompt Pattern

The **Data Assessment** pattern helps you understand your dataset before processing:

```
ROLE: Data science consultant
AUDIENCE: Materials science researcher
GOAL: Assess [dataset] for processing challenges and tool recommendations
INPUTS: [Specific dataset details]
TASK: Identify potential issues and recommend processing approach
CONSTRAINTS: [Word limits, time estimates, tool options]
EVAL CRITERIA: [Success metrics]
```

**Note**: The core RAG-IT components (Role, Audience, Goal, Inputs, Task) are essential. CONSTRAINTS and EVAL CRITERIA are additional components that help get better results.

#### 3. Iteration for Deeper Insights

After getting an initial AI assessment, you can iterate by:
- **Refining the prompt** with more specific details
- **Asking follow-up questions** about specific aspects (new or confusing from response)
- **Requesting validation** of your own analysis
- **Exploring edge cases** and potential problems

**Why This Matters**:
- **Efficiency**: AI can quickly identify data quality issues and processing challenges
- **Expertise**: AI provides domain-specific recommendations for materials science
- **Learning**: Iteration helps you develop deeper understanding of your data
- **Quality**: Better data understanding leads to better analysis results
