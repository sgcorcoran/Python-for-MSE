### Common Types of AI Errors in Materials Science (10-15 min)

#### 1. **Code Errors**

**Syntax Errors**: Invalid Python syntax that prevents code from running
- Missing colons `:` after if/for statements, missing parentheses `()` or brackets `[]`
- Incorrect indentation (Python is very sensitive to this!) - **Note**: This is easy to get messed up when copying and pasting Python code from AI output. Always check indentation after pasting code from Cursor.
- Missing quotes around strings

**Logic Errors**: Code runs but does the wrong thing
- Wrong function for the task (e.g., using `mean()` when you need `median()`)
- Incorrect calculations, missing steps in the analysis workflow

**Import/Function Errors**: Missing or incorrect library usage
- Using functions that don't exist, wrong parameters, missing import statements, deprecated functions

**Example**: AI generates `np.mean(data)` when `data` is a DataFrame and you are wanting to take the mean of just one column. Should be `data['column'].mean()` or `np.mean(data['column'])`

---


#### 2. **Statistical Errors**

**Wrong Test Selection**: AI recommends tests that don't match your data (e.g., t-test when data violates normality, parametric tests for non-normal data, wrong test for research question)

**💡 Reminder**: When we say "normal" or "normality" in statistics, we just mean the data follows a normal distribution (bell-shaped curve). 

**Missing Assumption Checks**: AI forgets to check test requirements (not checking normality, equal variances, or sample sizes)

<sup>1</sup> **Footnote**: Parametric tests (like t-test, ANOVA) assume your data follows a normal distribution (described by **parameters** such as mean and standard deviation) and use formulas that only work correctly when data is normally distributed. If your data is NOT normal (non-normal), these parametric tests give incorrect results. That's why you need **non-parametric** tests (like Mann-Whitney U, Kruskal-Wallis) for non-normal data - they don't assume any particular distribution parameters or shape (they're "distribution-free").

**Incorrect Interpretation**: AI misunderstands what the test means (confusing correlation with causation, overstating significance, ignoring practical vs. statistical significance)

**Example**: AI recommends t-test for two groups without checking if data is normal or if variances are equal

#### 3. **Factual Errors (Hallucinations)**

AI can make up information that sounds plausible: inventing property values (e.g., "aluminum yield strength is 500 MPa" when typical is 275-310 MPa), citing papers that don't exist, using incorrect formulas, mixing up units (MPa vs GPa, Celsius vs Fahrenheit)

**Example**: AI states that 7075 aluminum has yield strength of 150 MPa (actually ~470 MPa)

#### Step 3: Logic Validation (15-20 minutes)

### Systematic Debugging Workflow: The 4-Step Process

Before we start, let's learn a structured approach to debugging. This workflow will help you find and fix errors more efficiently.

#### The 4-Step Debugging Process:

**1. Isolate** (5-10 min)
- Identify which part of code is failing
- *Comment out working sections*
- Test smallest possible unit
- **Goal**: Find exactly where the problem is

**2. Identify** (10-15 min)
- Read error message carefully (if code runs)
- Check error type: syntax, logic, data, or environment?
- *Use print statements to inspect variables*
- Compare to working examples
- **Goal**: Understand what's wrong and why

**3. Fix** (10-20 min)
- Apply appropriate fix based on error type
- *Test fix with simple case first*
- Verify fix doesn't break other parts
- **Goal**: Make the code work correctly

**4. Verify** (5-10 min)
- *Run full code with test data*
- Check results make sense
- Validate against domain knowledge
- **Goal**: Confirm everything works and results are correct

#### Common Error Types and How to Spot Them:

- **Syntax Errors**: Code won't run, Python shows error immediately (missing colons, parentheses, quotes)
- **Logic Errors**: Code runs but gives wrong results (wrong function, incorrect calculation, missing steps)
- **Data Errors**: Code fails when data is wrong format/type (wrong column names, missing data, wrong data types)
- **Environment Errors**: Code fails due to missing libraries/imports (missing `import` statements, wrong library versions)

**💡 For Python Beginners**: Don't worry if you don't catch every error! The goal is to practice using this systematic approach. Let's work through the next example together.

### 📚 Worked Example: Using the Debugging Workflow (Practice)

**Before you start finding errors on your own, let's walk through one example using the 4-step debugging workflow!**

**Example Code (with one error):**

```python
# Calculate mean hardness
data = pd.read_csv('heat_treatment_data.csv')
mean_hardness = data.mean('Hardness_HV')  # ERROR: Wrong function call!
```

**Step-by-Step Debugging Process Using Our 4-Step Workflow:**

#### Step 1: Isolate (Find where the problem is)
- **What is this code trying to do?** Calculate mean hardness from a CSV file
- **Which line is problematic?** The `mean_hardness = data.mean('Hardness_HV')` line
- **Isolate the issue**: The problem is specifically with how we're calling the `mean()` function

#### Step 2: Identify (Understand what's wrong)
- **What type of error is this?** Logic error (the syntax looks correct, but the function is used incorrectly)
- **Check the function call**: Is `data.mean('Hardness_HV')` correct?
  - Let's check: In pandas, `DataFrame.mean()` calculates mean for all columns, not a specific column
  - To get mean of a specific column, we need: `data['Hardness_HV'].mean()`
- **Why it's a problem**: `mean()` on a DataFrame doesn't take a column name as a parameter this way
- **What happens**: If you run this code, you'll get an error like `ValueError: No axis named Hardness_HV for object type DataFrame` or `KeyError: 'Hardness_HV'` - this tells you the function call is incorrect

#### Step 3: Fix (Apply the correction)
- **How to fix**: Use `data['Hardness_HV'].mean()` instead
- **Correct code**:
  ```python
  data = pd.read_csv('heat_treatment_data.csv')
  mean_hardness = data['Hardness_HV'].mean()  # CORRECT!
  ```

#### Step 4: Verify (Confirm it works)
- **Test with simple case**: Run the corrected code
- **Check results**: Does the mean make sense? (For aluminum T6, hardness should be ~150-170 HV)
- **Validate understanding**: 
  - In pandas, to calculate the mean of a specific column:
    1. Select the column: `data['Hardness_HV']`
    2. Calculate mean: `.mean()`
    3. Combined: `data['Hardness_HV'].mean()`

**💡 Why This Matters**: This error is subtle - the code might run but give wrong results. Using the systematic debugging workflow helps you catch these errors before they lead to incorrect scientific conclusions!

### Statistical Concepts Primer (10 min)

**Before we validate AI statistical recommendations, let's briefly review the key concepts:**

#### Key Statistical Concepts (also look back to lesson 04 week 7 & 8):

**1. Normality**
- **What it means**: Data follows a bell-shaped curve (normal distribution)
- **Why it matters**: Many statistical tests assume data is normal
- **How to check**: 
  - Statistical test: Shapiro-Wilk test (`stats.shapiro()`)
  - Visual check: Histogram or Q-Q plot
- **Simple interpretation**: If p-value > 0.05, data is approximately normal

**2. Equal Variances**
- **What it means**: Two groups have similar spread (variability)
- **Why it matters**: Standard t-test assumes equal variances
- **How to check**: 
  - Statistical test: Levene's test (`stats.levene()`) or Bartlett's test (`stats.bartlett()`)
  - Visual check: Compare boxplot spreads
- **Simple interpretation**: If p-value > 0.05, variances are approximately equal

**3. Parametric vs Non-parametric Tests**
- **Parametric tests**: Assume normal distribution (t-test, ANOVA)
  - Examples: `stats.ttest_ind()` (two groups), `stats.f_oneway()` (multiple groups)
  - Use when: Data is normal AND assumptions are met
- **Non-parametric tests**: No distribution assumptions (Mann-Whitney U, Kruskal-Wallis)
  - Examples: `stats.mannwhitneyu()` (two groups), `stats.kruskal()` (multiple groups)
  - Use when: Data is NOT normal OR assumptions are not met
  - **What other assumptions?** For parametric tests like t-test, you also need equal variances between groups. If variances are unequal AND data is normal, you can use Welch's t-test instead. If other assumptions fail (like having very small sample sizes), use non-parametric tests.
- **Decision rule**: If data is normal → use parametric; if not normal → use non-parametric

**4. P-values Simplified**
- **What it means**: Probability of observing our two data sets if there's was no real difference in the sets.
- **Interpretation**: p < 0.05: Assuming our two groups of data have no difference, we can expect to collect similar data (or more extreme) as our two groups only 5% of the time. Therefore, we reject the null hypothesis (that they are the same) and conclude they are statistically different. 
- **Warning**: Small p-value doesn't mean large practical difference
- **Example**: p = 0.001 means "very unlikely this happened by chance" (strong evidence of difference)

**5. Statistical vs Practical Significance**
- **Statistical significance**: p < 0.05 (statistical test says there's a difference)
- **Practical significance**: Is the difference meaningful in real-world terms?
- **Example**: Mean hardness difference of 2 HV might be statistically significant but not practically meaningful (measurement error might be larger)

**💡 Decision Tree for Choosing Statistical Tests:**

```
Are you comparing two groups?
├─ Yes: Is data normal?
│   ├─ Yes: Are variances equal?
│   │   ├─ Yes: Use t-test (stats.ttest_ind)
│   │   └─ No: Use Welch's t-test (stats.ttest_ind with equal_var=False)
│   └─ No: Use Mann-Whitney U test (stats.mannwhitneyu)
└─ No: [Multiple groups - not covered in this lesson]
```

**💡 Remember**: These concepts help you validate AI recommendations. If AI recommends a parametric test but your data isn't normal, that's a red flag!
