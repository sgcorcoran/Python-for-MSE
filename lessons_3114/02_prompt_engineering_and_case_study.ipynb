### Mini‑Lecture (~25–30 min)

Prompt engineering is the art and science of communicating effectively with AI tools to get the best possible results. In materials science, this means learning how to ask AI tools the right questions to help with research planning, data analysis, and problem-solving.

**Core Principles**  
1. **Context** — Provide role, audience, and domain specifics (e.g., *aerospace, marine corrosion, tensile strength targets*).  
2. **Constraints** — Bound the task (word/format limits, data sources, property ranges, cite uncertainty).  
3. **Iteration** — Evaluate output; refine the next prompt with specific feedback (what to keep/change).  

**Prompt Anatomy Template**  
```
ROLE: [who is answering]
AUDIENCE: [who will use the output]
GOAL: [what you need]
INPUTS: [data/assumptions you provide]
CONSTRAINTS: [word/format, units, bounds, sources]
TASK: [the action: explain, compare, plan, compute, critique]
EVAL CRITERIA: [what “good” looks like]
```

**Example — Weak vs Strong**  

**Weak:**  
> "Explain alloy optimization."  

**Strong (materials context + constraints):**  

>ROLE: Materials engineer. AUDIENCE: Undergraduate students in MSE.  
GOAL: Outline 3 strategies to optimize an **Al–Mg** alloy for **marine corrosion resistance** while maintaining **yield strength ≥ 250 MPa**.  
INPUTS: Assume typical Mg 3–5 wt% and exposure to chloride-rich seawater.  
CONSTRAINTS: ≤180 words; include 1 trade‑off per strategy; use SI units.  
TASK: Provide bullet points with short rationales and expected property impacts.  
EVAL CRITERIA: Actionable, physically plausible, mentions microstructure/corrosion mechanisms.




### Worked Example 1: Iterative Improvement

**Prompt v1**  
```
ROLE: Materials engineer
GOAL: Improve marine corrosion resistance of Al–Mg without big strength loss
CONSTRAINTS: ≤120 words
TASK: List feasible adjustments
```
**Typical Output Issues to Watch For**  
- Vague mechanisms (e.g., "improve corrosion" without saying *how*).  
- No mention of **microstructure**, **precipitates**, or **chloride pitting**.  
- No **units** or **magnitudes** for property changes.  

**Prompt v2 (Refined)**  
```
ROLE: Materials engineer
AUDIENCE: Process team
GOAL: Increase pitting resistance in Al–Mg (3–5 wt% Mg), target YS ≥ 250 MPa
INPUTS: Marine, chloride exposure, T6-like aging possible
CONSTRAINTS: ≤150 words; SI units; note mechanism + trade‑off
TASK: Recommend 3 strategies; for each: (a) mechanism (grain size/precipitates/passivation), 
(b) process route (e.g., cold work + aging window), (c) expected ∆YS/∆corrosion qualitatively
EVAL CRITERIA: Specific, mechanism‑anchored, process‑aware
```
**Why v2 is better**  
- Names **mechanisms** and **processing windows** → more actionable.  
- Specifies **targets** and **constraints** → tighter, comparable answers.

### Workflow & Timeboxing

**Step 1 — Problem Framing (25–30 min)**  
- Define targets, constraints, validation metrics.  
- Draft **1–2 prompts** that set ROLE/AUDIENCE/GOAL/INPUTS/CONSTRAINTS/TASK/EVAL.  

**Step 2 — AI‑Assisted Reasoning (60–75 min)**  
- Run 2+ iterations. After each, annotate: *what to keep/change* next time.  
- Ask for **mechanism** (grain size, precipitates, passive film, pitting), **process route** (solution + aging windows), and **expected property impact**.  

**Step 3 — Validation Plan (25–35 min)**  
- Propose a **simple test matrix** (e.g., factor levels, sample count, metrics).  
- Include at least one **corrosion** metric and one **mechanical** metric.  

**Step 4 — Limits & Misuse Reflection (20–30 min)**  
- Where did AI hand‑wave or hallucinate? What data would you gather?  
- Ethical/safety/compliance concerns? (e.g., alloying additions, processing temps).

### Step 3: Create a research validation plan

1. **Ask the AI**: "Propose a simple test matrix (factors, levels, responses) for validating our alloy optimization strategy."
2. **Paste the output here.**
3. **Reformat** the AI's suggestion into a Python dictionary (see example below) so you can clearly see factors, levels, and responses to run.

Example format:
```python
plan = {
    "factors": ["Aging temp (°C)", "Aging time (h)", "Cold work (%)"],
    "levels": [["160", "180"], ["4", "8"], ["0", "10"]],
    "responses": ["Pitting rate (mm/yr)", "E_corr (V vs Ag/AgCl)", "YS (MPa)", "Hardness (HV)"]
}
```


🔎 **Factors, Levels and Responses** are common terminology in Design of Experiments (DOE)

>**Factors** = independent variables (what you change).  
**Levels** = settings for each factor.  
**Responses** = dependent variables (what you measure/observe).  

In a lab/science context, “responses” = “measurements” = “outputs you track.”

````python
# === Step 3: Validation Plan (Test Matrix) ===
# **** Note: This is an example plan. Edit it to match your specific scenario.**
plan = {
    "factors": ["Aging temp (°C)", "Aging time (h)", "Cold work (%)"],
    "levels": [["160", "180"], ["4", "8"], ["0", "10"]],
    "responses": ["Pitting rate (mm/yr)", "E_corr (V vs Ag/AgCl)", "YS (MPa)", "Hardness (HV)"]
}

print("Proposed factors & levels:\n")
for factor, levels in zip(plan["factors"], plan["levels"]):
    print(f"- {factor}: {', '.join(levels)}")

print("\nMeasurements needed:")
for resp in plan["responses"]:
    print(f"- {resp}")
````
