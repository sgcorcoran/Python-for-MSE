## Course Description & Learning Objectives

MSE 3114 introduces materials science students to the strategic integration of AI tools in research and analysis. Building upon Python programming from MSE 2114, students learn practical applications of AI tools (ChatGPT, Claude, GitHub Copilot) combined with modern data science techniques.

**By the end of this course, students will be able to:**
1. **Set up and configure AI tools** for materials science research
2. **Apply prompt engineering principles** for effective AI communication
3. **Utilize modern data science tools** (Polars, Streamlit, Plotly) for materials data
4. **Conduct statistical analysis** with AI assistance for interpretation
5. **Design experiments** using AI-guided optimization and DOE principles
6. **Implement machine learning models** for materials property prediction with AI assistance
