## 🚀 Step-by-Step Setup Instructions

### Step 1: Create the Main Project Folder

1. **Open File Explorer** (Windows key + E)
2. **Navigate to your desired location** - **DETAILED INSTRUCTIONS BELOW**
3. **Create a new folder:**
   - Right-click in empty space → "New" → "Folder"
   - Name it: `Python-for-mse`
   - Press Enter to confirm

### Step 2: Navigation Help - Don't Get Lost! 🗺️

**IMPORTANT:** Before you start clicking around, let's make sure you know where you are and where you're going!

#### Option A: Use the Address Bar (EASIEST METHOD)
1. **Look at the top of File Explorer** - you'll see a long white box with a path like `This PC > Documents`
2. **Click in that white box** (the address bar) - it will show the full path like `C:\Users\YourName\Documents`
3. **Type one of these paths** (replace `YourName` with your actual username):
   - `C:\Users\YourName\Documents` (for Documents folder)
   - `C:\Users\YourName\OneDrive` (for OneDrive folder)
   - `C:\Users\YourName\Desktop` (for Desktop folder)
4. **Press Enter** - File Explorer will take you directly there!

#### Option B: Use the Left Sidebar (Step-by-Step)
1. **Look at the left side** of File Explorer - you'll see a list of folders
2. **Click on "This PC"** first (this shows you all your drives)
3. **Click on "Local Disk (C:)"** (this is your main hard drive)
4. **Click on "Users"** folder
5. **Click on your username folder** (it will have your name)
6. **Click on "Documents"** (or "OneDrive" or "Desktop" - your choice!)

#### Option C: Quick Access (If Available)
1. **Look at the left sidebar** for "Quick access" section
2. **Click on "Documents"** or **"OneDrive"** if you see them listed there

#### 🚨 How to Know You're in the Right Place:
- **Check the address bar** at the top - it should end with something like:
  - `...\Documents` 
  - `...\OneDrive`
  - `...\Desktop`
- **Look at the folder contents** - you should see other folders like "Pictures", "Music", etc.

#### 🆘 If You Get Lost:
1. **Look at the address bar** - it shows exactly where you are
2. **Click the back arrow** (←) to go back one step
3. **Or click "This PC"** in the left sidebar to start over
4. **Or type `C:\Users\YourName\Documents`** in the address bar and press Enter

#### 💡 Pro Tip - Pin Your Location:
Once you find your Documents (or OneDrive) folder:
1. **Right-click on it** in the left sidebar
2. **Select "Pin to Quick access"**
3. **Now it will always appear** at the top of the left sidebar for easy access!

### Step 3: Create the Subfolder Structure

1. **Open the `Python-for-mse` folder** you just created
2. **Create the `lessons_3114` folder:**
   - Right-click in empty space → "New" → "Folder"
   - Name it: `lessons_3114`
3. **Create the `data_files` folder:**
   - Right-click in empty space → "New" → "Folder"
   - Name it: `data_files`
4. **Open the `data_files` folder** you just created
5. **Create the `database` folder:**
   - Right-click in empty space → "New" → "Folder"
   - Name it: `database`
   - Click on the address bar you should see `C:\Users\YourName\Documents\Python-for-mse\datafiles\database`

### Step 4: Download Files from Canvas

1. **Log into Canvas** and navigate to your MSE 3114 course
2. **Find the Files section** or the specific assignment/module
3. **Download the required files:**
   - **Lesson notebooks** (`.ipynb` files) → Save to `lessons_3114/` folder
   - **Data files** (`.csv` files) → Save to `data_files/` folder
   - **Data files** (`.json` files) → Save to `data_files/database/` folder
   - **Python scripts** (`.py` files) → Save to `lessons_3114/` folder

### Step 5: Organize Downloaded Files

1. **Check your Downloads folder:**
   - Press Windows key + R
   - Type: `%USERPROFILE%\Downloads`
   - Press Enter
2. **Move files to correct locations:**
   - **Select all `.ipynb` and `.py` files** → Cut (Ctrl+X) → Navigate to `Python-for-mse/lessons_3114/` → Paste (Ctrl+V)
   - **Select all `.csv` files** → Cut (Ctrl+X) → Navigate to `Python-for-mse/data_files/` → Paste (Ctrl+V)
   - **Select all `.json` files** → Cut (Ctrl+X) → Navigate to `Python-for-mse/data_files/database/` → Paste (Ctrl+V)

### Step 6: Open Project in Cursor

1. **Open Cursor** (if not already open)
2. **Open the project folder:**
   - Click "File" → "Open Folder" (or Ctrl+K, Ctrl+O)
   - Navigate to your `Python-for-mse` folder
   - Select the `Python-for-mse` folder (not a subfolder)
   - Click "Select Folder"

3. **Verify the structure in Cursor:**
   - You should see the folder structure in the left sidebar
   - Both `lessons_3114/` and `data_files/` folders should be visible
   - Your downloaded files should appear in their respective folders

## 🔧 Troubleshooting

### Common Issues:
1. **Files not showing in Cursor:** Refresh the file explorer (F5) or restart Cursor
2. **Wrong file locations:** Use File Explorer to move files to correct folders
3. **Permission errors:** Make sure you have write access to the folder location
4. **Canvas download issues:** Try downloading files one at a time instead of bulk download

### Quick File Paths:
- **Your Downloads:** `C:\Users\[YourUsername]\Downloads\`
- **Documents:** `C:\Users\[YourUsername]\Documents\`
- **OneDrive:** `C:\Users\[YourUsername]\OneDrive\`
- **Course Files:** `C:\Users\[YourUsername]\Documents\Python-for-mse\`
- **Alernate Course File Location:** `C:\Users\[YourUsername]\OneDrive\Python-for-mse`
- **3114 Lesson Files:** `C:\Users\[YourUsername]\Documents\Python-for-mse\lessons_3114\`
