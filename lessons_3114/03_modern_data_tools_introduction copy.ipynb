### Mini-Lecture (~25–30 min)

Modern data tools offer significant performance advantages over traditional pandas, but choosing the right tool requires understanding your specific materials science requirements. AI can help guide these decisions by analyzing your data characteristics, performance needs, and research goals.

**Core Principles for AI-Guided Tool Selection**  
1. **Data-First Analysis** — Let AI assess your dataset characteristics before tool selection  
2. **Performance Requirements** — Use AI to define specific performance targets and constraints  
3. **Iterative Optimization** — Test AI recommendations, measure results, refine approach  
4. **Materials Science Context** — Ensure AI considers domain-specific requirements and constraints  

**AI Prompt Patterns for Tool Selection**  

**Pattern 1: Tool Assessment**
```
ROLE: Data engineering consultant
AUDIENCE: Materials science research team
GOAL: Recommend optimal data processing tools for [specific materials dataset]
INPUTS: [Dataset size, data types, analysis requirements, hardware constraints]
CONSTRAINTS: ≤200 words; include performance estimates; suggest 2-3 tool options
TASK: Analyze requirements and recommend processing approach
EVAL CRITERIA: Specific, performance-focused, considers materials science context
```

**Pattern 2: Performance Optimization**
```
ROLE: Performance optimization specialist
AUDIENCE: Research computing team
GOAL: Optimize data processing performance for [specific analysis task]
INPUTS: [Current approach, performance metrics, bottlenecks, requirements]
CONSTRAINTS: ≤250 words; suggest specific optimizations; include trade-offs
TASK: Identify bottlenecks and recommend improvements
EVAL CRITERIA: Specific, measurable, considers resource constraints
```

**Pattern 3: Workflow Design**
```
ROLE: Data pipeline architect
AUDIENCE: Materials engineering team
GOAL: Design optimal data processing pipeline for [research objective]
INPUTS: [Data characteristics, analysis goals, performance requirements]
CONSTRAINTS: ≤300 words; include step-by-step process; note potential bottlenecks
TASK: Create processing pipeline with tool recommendations
EVAL CRITERIA: Feasible, efficient, materials-science appropriate
```

**Modern Data Tools for Materials Science**

**Polars** — Fast, memory-efficient alternative to pandas
- **Best for**: Large datasets (>100MB), complex aggregations, memory-constrained systems
- **Materials applications**: Property databases, composition analysis, statistical processing
- **AI can help with**: Syntax conversion, optimization strategies, performance tuning

**Dask** — Parallel computing for larger-than-memory datasets
- **Best for**: Very large datasets (>1GB), distributed computing, complex workflows
- **Materials applications**: High-resolution microscopy, extensive property databases
- **AI can help with**: Workflow design, resource allocation, optimization strategies

**Streamlit** — Interactive dashboard creation
- **Best for**: Research communication, interactive data exploration, stakeholder presentations
- **Materials applications**: Property visualization, composition optimization, results sharing
- **AI can help with**: UI design, functionality planning, deployment strategies

**When AI Recommendations May Be Limited**
- **Hardware-specific constraints**: AI may not know your exact system capabilities
- **Domain-specific requirements**: May miss materials science-specific considerations
- **Performance trade-offs**: May not fully understand your specific priorities
- **Tool ecosystem**: May not be aware of all available options or recent updates
