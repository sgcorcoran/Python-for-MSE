## Course Description

MSE 3114 is an advanced, project-based course that introduces materials science students to the strategic integration of artificial intelligence tools in research and analysis. Building upon the Python programming foundation from MSE 2114, this course focuses on practical applications of AI tools (ChatGPT, Claude, GitHub Copilot, local LLMs) combined with modern data science techniques for materials science problems.

Students will learn to leverage AI assistance for data analysis, statistical testing, experimental design, machine learning, and research communication while maintaining scientific rigor and critical thinking skills.

## Course Learning Objectives

By the end of this course, students will be able to:

1. **Set up and configure AI tools** for materials science research, including ChatGPT, Claude, GitHub Copilot, and local LLMs
2. **Apply prompt engineering principles** to effectively communicate with AI tools for materials science applications
3. **Utilize modern data science tools** including Polars for efficient data processing and Streamlit for interactive dashboards
4. **Conduct statistical analysis** with AI assistance for interpretation and decision-making in materials research
5. **Design and analyze experiments** using AI-guided optimization and modern DOE principles
6. **Implement machine learning models** for materials property prediction and classification with AI assistance
7. **Integrate AI tools throughout the research workflow** from planning to presentation
8. **Evaluate and validate AI-generated insights** while maintaining scientific rigor
